Run the cells in order. The sample amounts are in HKD.

```python
import faircart
from faircart import load_cart, optimize, compare_topup

print("Version:", faircart.__version__)
print("Import location:", faircart.__file__)
examples = Path("examples") if Path("examples/shared.json").exists() else Path(".")
```

## Shared savings
An equal split would charge Amy more than her solo price and budget.


```python
plan = optimize(load_cart(examples / "shared.json"))
print(plan.to_text())
assert plan.total == 10800
assert plan.payments == (1661, 4154, 4985)
assert sum(plan.payments) == plan.total
```

## A lower personal budget


```python
capped = optimize(load_cart(examples / "capped.json"))
print(capped.to_text())
assert capped.payments == (1000, 4455, 5345)
```

## Two personal coupons


```python
split = optimize(load_cart(examples / "split.json"))
print(split.to_text())
assert split.total == 7000
assert split.all_in_one.total == 8500
assert len(split.orders) == 2
```

## Budget shortfall


```python
blocked = optimize(load_cart(examples / "impossible.json"))
print(blocked.to_text())
assert not blocked.feasible
assert blocked.payments is None
assert blocked.shortfall == 2800
```

## The free-shipping trap


```python
topup = compare_topup(load_cart(examples / "topup.json"), owner="amy", price="15.00")
print(topup)
assert topup["before_total"] == "70.00"
assert topup["after_total"] == "73.00"
assert topup["cash_change"] == "3.00"
```
